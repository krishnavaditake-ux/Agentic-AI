ollama models are trained to generate vector embeddings

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings=(
    OllamaEmbeddings(model="gemma:2b") ## default runs lamma2 model
)
```

```python
r1=embeddings.aembed_documents(
    [
        "alpha is first greek letter",
        "beta is seccond letter of greek alphabet"
    ]
)
r1
```

```python
r2=embeddings.aembed_query(['beta is seccond letter of greek alphabet'])
r2
```
