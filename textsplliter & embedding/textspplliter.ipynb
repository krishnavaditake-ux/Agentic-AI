```python
from langchain_community.document_loaders import PyPDFLoader
loaders=PyPDFLoader('1706.03762v7.pdf')
docs=loaders.load()
```

how to recursively split text into characters
.................required library --> langchain-text-splitters

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
final_document=text_splitter.split_documents(docs)       ## we can also use create_document insted of split_document when file is text else
##final_document
```

```python
print(final_document[0])
print(final_document[1])
```

charactr text splitter

```python
from langchain_text_splitters import CharacterTextSplitter
text_splitter=CharacterTextSplitter(separator="/n",chunk_size=500,chunk_overlap=50)
text_splitter.split_documents(docs)
```

HTML HEADER TEXT SPLLITER

```python
from langchain_text_splitters import HTMLHeaderTextSplitter

html_string = """
<h1>Artificial Intelligence</h1>
<p>AI is the simulation of human intelligence.</p>

<h2>Machine Learning</h2>
<p>ML is a subset of AI.</p>

<h2>Deep Learning</h2>
<p>Deep learning uses neural networks.</p>
"""
## we can also gvie url of website insted to html_string
## url=""


headers_to_split_on = [
    ("h1", "Header 1"),
    ("h2", "Header 2"),
    ("h3", "Header 3")
]

html_splitter = HTMLHeaderTextSplitter(headers_to_split_on=headers_to_split_on)
header_split = html_splitter.split_text(html_string)

print(header_split)
```

JASON DATA SPLITTER

```python
import json
import requests

json_data = requests.get("https://dummyjson.com/users").json()
```

```python
print(json_data)
```

```python
from langchain_text_splitters import RecursiveJsonSplitter
json_splitter=RecursiveJsonSplitter(max_chunk_size=100)
json_chunk = json_splitter.split_json(json_data)
print(json_chunk)
```

```python
for chunk in json_chunk[:3]:
    print(chunk)
```

```python
## can also produce documents output
docs=json_splitter.create_documents(texts=[json_data])
for docs_sample in docs[:2]:
    print(docs_sample)
```
